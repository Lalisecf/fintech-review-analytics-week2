1. Import Libraries Cell

```python
# Core libraries
import pandas as pd
import numpy as np
import re
from datetime import datetime

# The star of the show
from google_play_scraper import app, reviews, Sort

print("Libraries loaded successfully!")
```

2. Define Bank Applications Cell

```python
# Google Play application IDs

apps = {
    "Commercial Bank of Ethiopia": "com.combanketh.mobilebanking",
    "Bank of Abyssinia": "com.boa.boaMobileBanking",
    "Dashen Bank": "com.dashen.dashensuperapp"
}
```

3. Scraping Google Play Reviews

We collect:

- Review text
- Rating
- Review date
- Bank name
- Source

Target:
- Minimum 400 reviews per bank

```python
all_reviews = []

for bank_name, app_id in apps.items():

    print(f"Scraping reviews for {bank_name}...")

    result, continuation_token = reviews(
        app_id,
        lang='en',
        country='et',
        sort=Sort.NEWEST,
        count=500
    )

    print(f"Collected {len(result)} reviews")

    for review in result:

        all_reviews.append({
            "review_id": review.get("reviewId"),
            "review": review.get("content"),
            "rating": review.get("score"),
            "date": review.get("at"),
            "bank": bank_name,
            "source": "Google Play"
        })

print("Scraping completed!")
```

4. Create DataFrame Cell

```python
df = pd.DataFrame(all_reviews)

print("Dataset Shape:")
print(df.shape)

df.head()
```

```python
print("Missing Values:")

df.isnull().sum()
```

```python
duplicate_count = df.duplicated(subset=["review_id"]).sum()

print(f"Duplicate Reviews: {duplicate_count}")
```

## Data Quality Check

The dataset was checked for missing values and duplicate reviews.

- Missing Values: `0`
- Duplicate Reviews: `0`

This indicates that the scraped data is clean, complete, and ready for preprocessing and analysis.

5. Preprocessing Markdown

## Data Preprocessing

Preprocessing steps:

1. Remove duplicate reviews
2. Drop rows with missing review text or rating
3. Standardize dates
4. Keep required columns only

```python
#Remove Duplicates Cell
df = df.drop_duplicates(subset=["review_id"])

print("Shape after removing duplicates:")
print(df.shape)
```

```python
#Handle Missing Values Cell
df = df.dropna(subset=["review", "rating"])

print("Shape after removing missing values:")
print(df.shape)
```

## Duplicate and Missing Value Handling

After removing duplicate reviews, the dataset shape remained `(1500, 6)`, indicating that no duplicate records were found.

After handling missing values, the dataset shape also remained `(1500, 6)`, showing that there were no missing values in the `review` and `rating` columns.

This confirms that the dataset is clean and complete for further analysis.

```python
#Normalize Dates Cel
df["date"] = pd.to_datetime(df["date"]).dt.strftime("%Y-%m-%d")

df.head()
```

```python
#Keep Required Columns Cell
cleaned_df = df[[
    "review",
    "rating",
    "date",
    "bank",
    "source"
]]

cleaned_df.head()
```

6. Dataset Summary Cell

```python
print("Final Dataset Shape:")
print(cleaned_df.shape)

print("\nReviews Per Bank:")
print(cleaned_df["bank"].value_counts())
```

7. Save CSV Cell

```python
import os

os.makedirs("../data/raw", exist_ok=True)

cleaned_df.to_csv("../data/raw/fintech_reviews_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
