1. Sentiment Distribution

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/analyzed_reviews.csv")

sns.countplot(
    data=df,
    x="bank",
    hue="sentiment_label"
)

plt.title("Sentiment Distribution by Bank")
plt.xticks(rotation=15)
plt.show()
```

## Sentiment Distribution by Bank — Interpretation

The visualization compares the distribution of positive, neutral, and negative sentiments across reviews for Commercial Bank of Ethiopia (CBE), Bank of Abyssinia (BOA), and Dashen Bank.

### Key Findings

#### Commercial Bank of Ethiopia (CBE)
- CBE has the highest number of positive reviews among the three banks.
- Negative reviews are relatively low compared to positive feedback.
- This suggests that many users are satisfied with the app’s overall performance, usability, and services.
- However, the presence of neutral and negative reviews indicates that some users still experience issues such as login problems or transaction delays.

#### Bank of Abyssinia (BOA)
- BOA shows the highest number of neutral and negative sentiments.
- Positive reviews are lower compared to CBE and Dashen Bank.
- This may indicate inconsistent user experience and a higher level of customer dissatisfaction.
- Common concerns may include app crashes, slow performance, or failed transactions.
- The bank may need to prioritize bug fixes and service reliability improvements.

#### Dashen Bank
- Dashen Bank also receives a strong number of positive reviews, close to CBE.
- Negative reviews are moderate but lower than BOA.
- This suggests that users generally have a favorable experience with the application.
- Improvements in customer support responsiveness and transaction stability could further enhance user satisfaction.

### Overall Comparison
- Positive sentiment dominates across all three banking apps, indicating that mobile banking adoption and user engagement are generally strong.
- BOA appears to face more customer experience challenges compared to the other banks.
- CBE and Dashen Bank demonstrate relatively better customer satisfaction and app reliability.
- Negative and neutral reviews across all banks highlight recurring issues such as login errors, OTP delays, and slow transaction processing.

### Business Insight
The analysis suggests that improving transaction speed, login reliability, and customer support could significantly improve customer satisfaction and app ratings for all three banks. Banks with higher positive sentiment, particularly CBE and Dashen, should continue enhancing their strengths while addressing recurring technical complaints.

2. Theme Frequency

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    y="identified_theme",
    order=df["identified_theme"].value_counts().index
)

plt.title("Most Common Themes")
plt.show()
```

### Interpretation of the Theme Frequency Plot

The visualization shows the distribution of identified themes extracted from customer reviews of the three Ethiopian banking applications.

The most noticeable result is that the **"Other"** category dominates the dataset, containing the majority of reviews. This indicates that many customer comments did not match the predefined keyword-based themes. The large proportion of "Other" reviews suggests that user feedback is highly diverse and that the current keyword mapping could be expanded with additional keywords or more advanced topic modeling techniques such as LDA or clustering.

Among the classified themes, **Transaction Performance** appears most frequently. This suggests that users are highly concerned about issues related to transfers, payment speed, delays, and transaction reliability. Since banking applications are transaction-focused systems, this theme represents a critical business area requiring continuous optimization.

The **UI & UX** theme is also common, indicating that customers frequently discuss app design, navigation experience, and usability. Positive feedback in this category may reflect satisfaction with interface simplicity, while negative feedback may indicate frustrations with app layout or user interaction.

**Customer Support** and **Account Access Issues** appear at moderate levels. These themes reflect recurring concerns such as login failures, OTP verification problems, password recovery difficulties, and dissatisfaction with customer assistance responsiveness.

Finally, **Feature Requests** has the lowest frequency among the identified themes. This suggests that most users are currently more focused on operational reliability and problem resolution rather than requesting new functionalities.

### Business Implications

The findings indicate that Ethiopian mobile banking users prioritize:
- Reliable and fast transaction processing
- Smooth account access and authentication
- User-friendly application interfaces

Banks should therefore focus first on improving system stability, reducing transaction delays, and strengthening authentication reliability before introducing additional features.
